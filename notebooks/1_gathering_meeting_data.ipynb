```python
#To get a larger data set we're gonna add meetings from more years - all the way back to 2010,
#but skipping 2020-2022 because voting was remote during covid.
more_thursday_meetings = []
more_years_all_json_folder = "more_years_raw_all_meetings_jsons"
os.makedirs(more_years_all_json_folder, exist_ok=True)

# 2019 will be the real test year!
more_years = ["2019", "2018", "2017", "2016", "2015", "2014", "2013", "2012", "2011", "2010"]

for year in more_years:
    print(f"\n[{year}] Fetching data from API...")
    url = "https://data.europarl.europa.eu/api/v2/meetings"

    # Removed the limit parameter to match your working 2025 code!
    params = {
        "year": year,
        "format": "application/ld+json"
    }
    
    try:
        response = requests.get(url, headers=headers, params=params)
        if response.status_code == 200:
            data = response.json()
            try:
                safe_filename = f"meetings_{year}.json"
                file_path = os.path.join(more_years_all_json_folder, safe_filename)
                with open(file_path, "w", encoding="utf-8") as f:
                    json.dump(data, f, indent=4)
            except Exception as save_err:
                print(f"  -> Warning: Could not save file for {year}: {save_err}")

            thursday_count = 0
            
            if "data" in data and isinstance(data["data"], list):
                
                for meeting in data["data"]:
                    date_str = meeting.get("activity_start_date") or meeting.get("activity_date")
                    if not date_str and isinstance(meeting.get("eli-dl:activity_date"), dict):
                        date_str = meeting["eli-dl:activity_date"].get("@value")
                    
                    if date_str:
                        clean_date = date_str.replace("Z", "+00:00")
                        # Fixed: Passing clean_date into the format instead of date_str
                        date_obj = datetime.fromisoformat(clean_date) 
                        
                        if date_obj.weekday() == 3:
                            more_thursday_meetings.append(meeting)
                            thursday_count += 1
    
                print(f"[{year}] Filtered: Found {thursday_count} Thursdays.")
                        
        time.sleep(0.5)

    except Exception as e:
        print(f"Something went wrong: {e}")

df_more_thursdays = pd.DataFrame(more_thursday_meetings)
print(f"\n--- FETCH COMPLETE ---")
print(f"Total historical Thursday meetings found: {len(df_more_thursdays)}")

if not df_more_thursdays.empty:
    print(df_more_thursdays.head())
```
